# Pandas: El Motor del Análisis de Datos en Python


**Pandas** (estilizado frecuentemente como **pandas**) es una de las librerías de código abierto más importantes y utilizadas para el lenguaje de programación Python, diseñada específicamente para ofrecer **estructuras de datos de alto rendimiento, fáciles de usar y herramientas de análisis de datos**. Su nombre deriva del término econométrico _panel data_ (datos de panel), que se refiere a conjuntos de datos estructurados multidimensionales, y es un juego de palabras con la frase "Python data analysis".

Esta librería fue creada originalmente por **Wes McKinney** en 2008 para resolver la necesidad de una herramienta integrada que permitiera manipular datos financieros de manera flexible y potente, algo que no existía de forma completa en Python en aquel momento.

#### Estructuras de Datos Principales

Pandas se basa en dos objetos fundamentales que permiten organizar la información de manera similar a como lo haría una hoja de cálculo o una base de datos relacional:

**Series**: Es un objeto tipo arreglo **unidimensional** que contiene una secuencia de valores y un arreglo asociado de etiquetas de datos, llamado su **índice**. Se puede pensar en una Serie como un diccionario ordenado de longitud fija o una columna individual de una tabla.

**DataFrame**: Es la estructura más utilizada y representa una **tabla rectangular de datos**. Contiene una colección ordenada de columnas, donde cada una puede tener un tipo de valor diferente (numérico, cadena, booleano, etc.). El DataFrame tiene tanto un índice de filas como etiquetas de columnas, lo que lo hace similar a una tabla de SQL o una hoja de Excel.

#### ¿Cuál es su utilidad?

La utilidad de pandas es vasta en el campo de la **ciencia de datos** y el análisis estadístico, actuando como el "pegamento" que permite preparar los datos antes de aplicar modelos complejos. Sus funciones principales incluyen:

**Carga y almacenamiento de datos**: Pandas facilita la lectura de datos desde una gran variedad de formatos, incluyendo archivos **CSV**, archivos de texto delimitados, **Microsoft Excel**, archivos binarios (como HDF5 y Feather), documentos **HTML**, **JSON** y directamente desde **bases de datos SQL**.

**Limpieza y preparación de datos (Data Wrangling)**: Es su uso más crítico, permitiendo "limpiar" datos desordenados mediante el filtrado de valores, la eliminación de duplicados y, especialmente, el **manejo de datos faltantes** (representados como `NaN` o `NA`) a través de métodos para detectarlos, eliminarlos o rellenarlos.

**Manipulación y transformación**: Permite realizar operaciones de "cortar y pegar" (slice and dice), remodelar (reshape) y pivotar conjuntos de datos. También facilita la creación de **columnas derivadas** a partir de datos existentes.

**Agregación y operaciones de grupo**: Implementa el paradigma **split-apply-combine**, que permite dividir un conjunto de datos en grupos (basados en una o más claves), aplicar funciones estadísticas (como suma, media o desviación estándar) a cada grupo y combinar los resultados en una nueva tabla o reporte.

**Análisis de Series Temporales**: Posee herramientas extremadamente robustas para trabajar con datos indexados por tiempo, permitiendo realizar conversiones de frecuencia (resampling), desplazamientos temporales (shifting) y cálculos de ventanas móviles (rolling windows) para suavizar datos ruidosos.

**Integración con el ecosistema de Python**: Pandas se integra perfectamente con otras librerías fundamentales. Utiliza **NumPy** para realizar cálculos numéricos veloces y **Matplotlib** para generar visualizaciones directas de los datos contenidos en sus estructuras.

**Puente hacia el Machine Learning**: En los flujos de trabajo de aprendizaje automático, pandas es la herramienta estándar para la **ingeniería de características** (feature engineering), preparando las matrices de datos que luego se alimentan a librerías de modelado como **scikit-learn** o **statsmodels**.

### **1.1) Las Series en Pandas**

Una **Serie** es la estructura de datos más básica de pandas. Puedes imaginarla como un arreglo unidimensional (similar a una lista de Python o un array de NumPy), pero con una característica distintiva: tiene un **índice asociado**.

Mientras que en una lista común accedes a los elementos solo por su posición numérica (0, 1, 2...), en una Serie cada dato puede tener una etiqueta o nombre (índice). Esto permite que los datos estén "etiquetados", facilitando búsquedas y operaciones de alineación automática. Conceptualmente, una Serie es como un diccionario ordenado de longitud fija donde las claves son las etiquetas del índice y los valores son los datos.
Para crear una **Series de pandas**, primero debes asegurarte de tener instalada e importada la librería, lo cual se hace convencionalmente usando el alias `pd`. Una Series es una estructura de datos fundamental que consiste en un arreglo unidimensional que contiene una secuencia de valores y un arreglo asociado de etiquetas, llamado su **índice**.

Existen diversas formas de construir una Series dependiendo del origen de tus datos:

#### a. A partir de una lista o arreglo

La forma más básica es pasar una lista de valores al constructor `pd.Series()`. Si no especificas etiquetas adicionales, pandas creará automáticamente un índice por defecto compuesto por números enteros que van desde 0 hasta N−1 (donde N es la longitud de los datos).

```
import pandas as pd
obj = pd.Series([4, 7, -5, 3])
```
#### b. Usando etiquetas de índice personalizadas

En muchas aplicaciones, es preferible identificar cada dato con una etiqueta descriptiva en lugar de una posición numérica. Puedes lograr esto pasando una lista al argumento `index`:

```
obj2 = pd.Series([4, 7, -5, 3], index=["d", "b", "a", "c"])
```
Una ventaja de esto es que puedes usar estas etiquetas para seleccionar valores específicos de forma intuitiva.

#### c. A partir de un diccionario de Python

Puedes convertir un **diccionario** directamente en una Series. En este proceso, las claves del diccionario se convierten automáticamente en las etiquetas del índice y los valores del diccionario en los datos de la Series:

```
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}
obj3 = pd.Series(sdata)
```
Si además pasas una lista de etiquetas al parámetro `index` junto con el diccionario, pandas alineará los valores con las etiquetas proporcionadas. Si alguna etiqueta de tu lista no está presente en el diccionario, se le asignará el valor **NaN** (Not a Number), que es la forma en que pandas representa los **datos faltantes**.

#### d. Extrayendo columnas de un DataFrame

Un DataFrame es básicamente una colección de Series que comparten el mismo índice. Por lo tanto, si ya tienes un DataFrame, puedes crear una Series simplemente seleccionando una de sus columnas mediante su nombre entre corchetes.

```
# Suponiendo que 'df' es un DataFrame
serie_columna = df["nombre_columna"]
```
#### **Propiedades clave al crear Series**

**Alineación de datos:** Una característica potente de las Series es que se alinean automáticamente por etiqueta de índice cuando realizas operaciones aritméticas entre ellas.

**Nombres:** Tanto el objeto Series como su índice pueden tener un atributo `name`, lo cual es útil para identificar la información cuando se integra con otras funciones de pandas.

**Tipos de datos:** Aunque tradicionalmente se basan en arreglos de NumPy (como `float64` o `int64`), pandas también ofrece **tipos de extensión** más avanzados, como `Int64` (que permite valores nulos en enteros) o el tipo `string`, que es más eficiente para procesar texto.



### **1.2) Los DataFrames en Pandas**

El **DataFrame** es la estructura reina de pandas. Si la Serie es una columna, el DataFrame es la **tabla completa**. Es una estructura bidimensional (filas y columnas) donde cada columna puede tener un tipo de dato diferente (números, texto, booleanos, etc.).

Puedes visualizarlo como una hoja de cálculo de Excel o una tabla de base de SQL. Lo lógico es ver un DataFrame como un diccionario de Series que comparten el mismo índice. Posee tanto un índice de filas (eje 0) como un índice de columnas (eje 1). Esta estructura permite manipular datos complejos de forma masiva y eficiente, siendo el estándar para el análisis y la limpieza de datos en Python.

#### Ejemplos Contextualizados

**Ejemplo Teórico (Basado en el texto):** Un DataFrame de países "BRICS" que contenga las columnas "country", "capital", "area" y "population". Cada fila representa un país y cada columna una característica. Podemos acceder a la columna "area" para realizar estadísticas o filtrar las filas para ver solo países con una población mayor a cierto umbral.

**Ejemplo Práctico (Vida Real):** Un sistema de inventario de una tienda de electrónica. Las columnas serían "Producto", "Precio\_Unitario", "Cantidad\_Stock" y "Categoría". Cada fila sería un artículo diferente (ej. Laptop, Mouse). Si queremos saber el valor total del inventario, simplemente multiplicamos la columna "Precio\_Unitario" por "Cantidad\_Stock" para generar una nueva columna de "Valor\_Total".

Existen varias formas principales de crear un DataFrame dependiendo del origen de tus datos:

#### a. A partir de un diccionario de listas o arreglos

Esta es una de las formas más comunes de construcción. En este caso, las **claves del diccionario se convierten en las etiquetas de las columnas** y las listas (que deben tener la misma longitud) se convierten en los datos de dichas columnas.

```
import pandas as pd
datos = {
    "estado": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada"],
    "año": ,
    "pob": [1.5, 1.7, 3.6, 2.4, 2.9]
}
df = pd.DataFrame(datos)
```
Si deseas que las columnas aparezcan en un orden específico diferente al del diccionario, puedes indicarlo con el argumento `columns`.

#### b. A partir de una lista de diccionarios

En este método, el DataFrame se construye **fila por fila**. Cada diccionario de la lista representa una fila, donde las claves del diccionario actúan como los nombres de las columnas.

```
lista_dict = [
    {"nombre": "Ginger", "raza": "Dachshund", "peso_kg": 10},
    {"nombre": "Scout", "raza": "Dalmatian", "peso_kg": 25}
]
df = pd.DataFrame(lista_dict)
```
#### c. A partir de un diccionario de diccionarios (Anidados)

Si pasas un diccionario anidado al constructor, pandas interpretará las **claves externas como los nombres de las columnas** y las **claves internas como los índices de las filas**.

```
poblaciones = {
    "Ohio": {2000: 1.5, 2001: 1.7},
    "Nevada": {2001: 2.4, 2002: 2.9}
}
df = pd.DataFrame(poblaciones)
```
#### d. A partir de arreglos de NumPy

Puedes crear un DataFrame pasando un **arreglo bidimensional (2D ndarray)** de NumPy. Puedes asignar etiquetas personalizadas para las filas y las columnas mediante los argumentos `index` y `columns` respectivamente; de lo contrario, se asignarán números por defecto.

```
import numpy as np
matriz = np.arange(9).reshape((3, 3))
df = pd.DataFrame(matriz, columns=["A", "B", "C"], index=["f1", "f2", "f3"])
```
#### f. Cargando datos desde archivos externos

Pandas es extremadamente eficiente para crear DataFrames a partir de archivos de datos existentes:

**Archivos CSV:** Es el formato más utilizado. Se crea mediante `pd.read_csv('ruta/archivo.csv')`.

**Archivos Excel:** Se utiliza `pd.read_excel('archivo.xlsx')` tras instalar librerías adicionales como `openpyxl`.

**JSON:** Ideal para datos web, se cargan con `pd.read_json()`.

#### g. Propiedades clave al crear un DataFrame

**Índice de filas:** Si no especificas uno al crear el DataFrame, pandas asignará automáticamente un índice numérico entero que comienza en 0 hasta N−1. Puedes establecer cualquier columna como índice después de la creación usando el método `.set_index()`.

**Inspección inicial:** Una vez creado, puedes usar el método `.head()` para visualizar las primeras 5 filas o `.info()` para verificar los tipos de datos de las columnas y la existencia de valores nulos.

**Valores faltantes:** Si al crear el DataFrame a partir de un diccionario faltan datos para ciertas etiquetas de fila, pandas insertará automáticamente el valor **NaN** (Not a Number) para representar la ausencia de datos.

## **2) Carga, Almacenamiento y Formatos de Archivos**

La carga de datos, a menudo denominada "parsing" o análisis sintáctico, representa el pilar fundamental de cualquier flujo de trabajo en la ciencia de datos. Este proceso consiste en leer información almacenada en archivos externos y transformarla en estructuras que Python pueda manipular, principalmente DataFrames de pandas. Debido a que los datos en el mundo real suelen ser "sucios" o estar mal estructurados, las herramientas de carga deben ser extremadamente flexibles.

Podemos clasificar los formatos de archivos en tres grandes grupos:

-   **Formatos de Texto Plano:** Son los más comunes debido a su portabilidad. Los archivos **CSV (valores separados por comas)** y otros formatos delimitados son archivos de texto que cualquier sistema puede leer. Pandas utiliza inferencia de tipos para decidir automáticamente si una columna contiene números, fechas o cadenas de texto. Sin embargo, en archivos muy grandes, esta flexibilidad requiere parámetros específicos para manejar encabezados faltantes, saltar filas de metadatos o identificar marcadores personalizados de valores faltantes (como "N/A" o "?").

-   **Formatos Estructurados y Web:** El formato **JSON** es el estándar para el intercambio de datos en la web y aplicaciones móviles. A diferencia de las tablas rígidas, permite estructuras anidadas similares a los diccionarios de Python. Por otro lado, los formatos **XML y HTML** permiten extraer datos directamente de páginas web o archivos de configuración jerárquicos.

-   **Formatos Binarios y Eficientes:** Para el almacenamiento a largo plazo o el manejo de grandes volúmenes de datos ("Big Data"), los formatos de texto son ineficientes. Formatos como **Pickle** (nativo de Python), **HDF5** (ideal para datos científicos masivos) y **Apache Parquet** ofrecen una velocidad de lectura y escritura drásticamente superior y una mayor compresión de espacio.

Finalmente, la integración con **bases de datos SQL** mediante librerías como SQLAlchemy permite que pandas ejecute consultas directamente y reciba los resultados listos para el análisis, cerrando el ciclo entre el almacenamiento masivo y el procesamiento interactivo.

A continuación se presentan ejemplos detallados y contextos comunes de uso para las funciones de carga, inferencia y almacenamiento de datos en pandas, conforme a lo descrito en las fuentes disponibles:

### **2.1) Lectura de Datos (Parsing)**

La carga de información es el primer paso esencial en el análisis de datos, proceso también conocido como **parsing** o análisis sintáctico. Pandas ofrece funciones altamente flexibles para convertir archivos externos en estructuras operativas como los DataFrames:

- `read_csv`: Es la función más utilizada para cargar archivos de texto delimitados por comas. Un ejemplo común es la carga de datos directamente desde una **URL**, como se observa con datasets de COVID-19 (`pd.read_csv('https://www.fdsp.net/data/covid-merged.csv')`). También se usa para datasets locales como el del Titanic o nombres de bebés. Admite parámetros críticos como `index_col` para definir qué columna será el índice y `skiprows` para ignorar líneas de encabezados no deseados.

- `read_table`: Se emplea habitualmente cuando el delimitador no es una coma, sino otro carácter o patrón. Por ejemplo, para procesar el dataset de **MovieLens** donde los campos están separados por dos puntos dobles (`::`), se utiliza `pd.read_table(path, sep="::", engine="python")`.

- `read_excel`: Permite importar datos de hojas de cálculo de Microsoft Excel (.xls o .xlsx). Es común especificar el nombre de la hoja mediante el argumento `sheet_name` para extraer solo la pestaña de interés.

- `read_json`: Utilizada para leer el formato estándar de intercambio de datos en la web (JavaScript Object Notation). Un caso de uso frecuente es convertir arreglos de objetos JSON donde cada objeto representa una fila de una tabla.

- `read_html`: Útil para el **web scraping**, ya que busca automáticamente etiquetas `<table>` dentro de un documento HTML y las devuelve como una lista de DataFrames. Un ejemplo es la extracción de tablas de fallos bancarios desde sitios gubernamentales.

- `read_sql`: Facilita la integración con **bases de datos relacionales** (como SQLite o PostgreSQL). Generalmente se usa junto con librerías como **SQLAlchemy** para ejecutar una consulta SQL y cargar los resultados listos para el análisis (`pd.read_sql("SELECT * FROM test", db)`).

### **2.2) Inferencia de Tipos**

Pandas realiza una **inferencia de tipos automática** al cargar archivos de texto (como CSV o TXT) que no incluyen metadatos sobre sus columnas.

Durante la lectura, la librería analiza los valores para decidir si una columna debe ser tratada como **numérica, booleana, de fecha o de texto**.

Sin embargo, esta inferencia puede ser ajustada manualmente. Por ejemplo, se puede forzar que una columna de números se lea como texto o utilizar el tipo `Int64` (extensión de pandas) para permitir valores nulos en columnas de enteros, evitando que se conviertan automáticamente a punto flotante (`float64`).

### **2.3) Escritura de Datos (Serialization)**

La **serialización** permite exportar los resultados del procesamiento a disco para su uso posterior o para compartir la información.

- `to_csv`: Exporta los datos a archivos de texto delimitados. Es común usar `index=False` para evitar que pandas escriba una columna adicional con los números de fila, y `na_rep` para definir cómo deben representarse los valores nulos en el archivo resultante (por ejemplo, escribiendo "NULL" en lugar de dejar el espacio vacío).

- `to_excel`: Guarda el DataFrame en formato Excel. Si se necesitan guardar múltiples DataFrames en un mismo archivo, se utiliza la clase `ExcelWriter` para manejar las diferentes hojas.

- `to_json`: Convierte el objeto a una cadena JSON o lo escribe en un archivo. El argumento `orient="records"` es un ejemplo común para exportar los datos como una lista de diccionarios, formato ideal para APIs web.

- `to_pickle`: Guarda objetos en el formato nativo de Python (**pickle**). Se recomienda principalmente para **almacenamiento a corto plazo**, ya que no se garantiza que el formato sea estable en versiones futuras de las librerías, aunque es extremadamente rápido para persistir estructuras complejas de Python.

### **2.4) Formatos Binarios**

Para manejar grandes volúmenes de datos o mejorar la eficiencia de entrada/salida (I/O), pandas soporta formatos binarios avanzados:

**HDF5 (vía** `HDFStore`**)**: Es un formato diseñado para almacenar cantidades masivas de datos científicos en arreglos. Permite la **compresión sobre la marcha** y es ideal cuando el dataset no cabe completamente en la memoria RAM, ya que permite leer y escribir secciones específicas del archivo de forma eficiente.

**Apache Parquet**: Este formato binario es el estándar para el **almacenamiento distribuido** (como en Amazon S3 o HDFS). Es altamente eficiente para consultas analíticas porque organiza los datos de forma columnar, lo que acelera drásticamente la lectura de columnas específicas en archivos de gran tamaño.

**Pickle**: Aunque es el formato más sencillo de usar para serializar cualquier objeto de Python, su uso se limita a entornos donde solo se utilice Python y para tareas que no requieran compatibilidad a largo plazo.

## **3) Exploración e Inspección de Datos**

La exploración e inspección de datos es la fase inicial y crítica de cualquier proyecto de ciencia de datos. Su propósito principal es permitir que el analista desarrolle una comprensión intuitiva y técnica del conjunto de datos antes de proceder a análisis más profundos o a la creación de modelos. Este proceso se puede dividir en tres dimensiones fundamentales: la inspección de estructura, el resumen estadístico y la visualización exploratoria.

En primer lugar, la **inspección estructural** implica revisar la arquitectura del archivo. Al cargar datos en herramientas como pandas, el primer paso es verificar la integridad de la carga mediante la visualización de las primeras filas. Esto permite confirmar que las columnas tienen los nombres correctos y que los tipos de datos (numéricos, cadenas de texto o fechas) se han interpretado adecuadamente. Es esencial identificar el tamaño del dataset y la presencia de valores faltantes (identificados comúnmente como `NaN`), ya que esto determinará las necesidades de limpieza posterior.

En segundo lugar, el uso de **estadísticas descriptivas** proporciona un resumen cuantitativo de las variables. Métricas como la media, la mediana, el valor mínimo, el máximo y la desviación estándar ofrecen una visión rápida de la centralidad y la dispersión de los datos. Esta etapa ayuda a entender el rango de los valores y a detectar inconsistencias lógicas (por ejemplo, pesos negativos o edades imposibles).

Finalmente, la **visualización exploratoria** es la herramienta más potente para identificar patrones que los números por sí solos podrían ocultar. Los diagramas de dispersión (scatter plots) son fundamentales para observar la relación entre dos variables y, muy especialmente, para detectar "outliers" o valores atípicos. Un valor atípico es aquel que se aleja significativamente del resto de las observaciones y que puede deberse a errores de entrada de datos, fallos de medición o descubrimientos genuinos de fenómenos inusuales. Por otro lado, los histogramas permiten observar la forma de la distribución de una variable, revelando si los datos siguen una distribución normal o si están sesgados hacia ciertos valores.

### 4.1) Indexación en pandas

La indexación es el proceso de asignar etiquetas a las filas de un DataFrame para facilitar su identificación y acceso. Por defecto, pandas asigna un índice numérico entero que comienza en 0, pero en ciencia de datos es común transformar una de las columnas (como nombres, fechas o IDs) en el índice principal mediante el método `set_index()`. Esto convierte a esa columna en el eje de referencia para las búsquedas.

Además, existe la **indexación jerárquica** o **MultiIndex**, que permite tener múltiples niveles de etiquetas en un mismo eje. Esto es extremadamente útil para representar datos de dimensiones superiores en una estructura bidimensional, permitiendo organizar los datos por categorías y subcategorías (por ejemplo, agrupar estados y luego años dentro de cada estado). El proceso inverso, `reset_index()`, devuelve las etiquetas al cuerpo del DataFrame como una columna normal, lo cual es útil cuando se necesita realizar operaciones que no admiten índices complejos.

### 4.2) Selección de datos

La selección se refiere a la extracción de partes específicas (columnas, filas o celdas) de un DataFrame. Para seleccionar columnas, se utiliza la notación de corchetes `df['NombreColumna']`, que devuelve una **Series**, o `df[['Col1', 'Col2']]`, que devuelve un **DataFrame**.

Para un control más preciso, pandas ofrece dos operadores fundamentales:

`.loc[]`: Selección basada en **etiquetas**. Permite acceder a filas y columnas mediante sus nombres. Por ejemplo, `df.loc['Fila1', 'ColumnaA']`.

`.iloc[]`: Selección basada en **posiciones enteras**. Se utiliza cuando conocemos la ubicación exacta en la matriz de datos, sin importar el nombre de la etiqueta. Por ejemplo, `df.iloc[0:5, 0:2]` seleccionaría las primeras 5 filas y las primeras 2 columnas.

Es importante entender que `.loc` incluye el punto final en los cortes (slices), mientras que `.iloc` sigue la convención estándar de Python de excluirlo.

#### ¿Qué diferencia hay entre usar .loc y .iloc para selección?

A continuación, se detallan las diferencias técnicas y de comportamiento entre ambos operadores:

#### a) Etiquetas frente a Posiciones

**.loc (Label-based):** Se utiliza para seleccionar datos mediante las **etiquetas de las filas y columnas**. Por ejemplo, si un DataFrame tiene nombres de estados como índice, se usaría `df.loc['Mendoza']` para acceder a esa fila específica.

**.iloc (Integer-position-based):** Se utiliza para acceder a los datos basándose en su **ubicación numérica** en la matriz, empezando desde 0 hasta la longitud menos uno. Por ejemplo, `df.iloc[2]` recuperará la tercera fila del DataFrame, independientemente de qué nombre tenga asignado su índice.

#### b) Comportamiento en el Slicing (Rebanado)

El comportamiento al seleccionar rangos (slices) varía significativamente entre ambos:

**Inclusividad en .loc:** Al realizar un recorte con etiquetas, **el punto final está incluido**. Si se solicita `df.loc['a':'c']`, el resultado incluirá tanto la fila 'a' como la 'c'.

**Exclusividad en .iloc:** Sigue la convención estándar de Python, donde **el punto final es exclusivo**. Al ejecutar `df.iloc[0:3]`, se seleccionarán las posiciones 0, 1 y 2, pero no la 3.

#### c) Selección Multidimensional

Ambos operadores permiten seleccionar subconjuntos de filas y columnas de forma simultánea utilizando una coma para separar los ejes (`[filas, columnas]`):

Con **.loc**, se pueden pasar listas de etiquetas para ambos ejes: `df.loc[['Fila1', 'Fila2'], ['ColumnaA', 'ColumnaB']]`.

Con **.iloc**, se utilizan listas de enteros para las posiciones: `df.iloc[2:5, 0:3]`.

#### d) Uso de Arreglos Booleanos

Existe una distinción importante en cuanto al filtrado lógico:

**Filtrado con .loc:** Permite el uso de **arreglos booleanos** para seleccionar filas que cumplan una condición específica. Por ejemplo, `df.loc[df['columna'] > 5]` es una operación válida.

**Restricción en .iloc:** A diferencia de `.loc`, el operador **.iloc no admite arreglos booleanos** para la selección.

### 4.3) Filtrado de DataFrames

El filtrado permite extraer filas que cumplen con condiciones lógicas específicas, una técnica conocida como **indexación booleana**. Al aplicar una comparación (como `df['Edad'] > 18`), pandas genera una serie de valores True/False. Al pasar esta serie al DataFrame original, este solo devuelve las filas donde el valor es True.

Podemos combinar múltiples condiciones utilizando los operadores lógicos:

`&` (AND): Ambas condiciones deben cumplirse.

`|` (OR): Al menos una condición debe cumplirse.

`~` (NOT): Invierte la condición.

Para filtros más legibles y complejos, pandas ofrece el método `.query()`, que permite escribir la condición como una cadena de texto (string), facilitando la integración de variables externas mediante el símbolo `@`. El filtrado es la herramienta principal para la limpieza de datos y la eliminación de valores atípicos (outliers).
