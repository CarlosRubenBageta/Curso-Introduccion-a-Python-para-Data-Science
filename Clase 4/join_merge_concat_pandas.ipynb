## 🔷 1. `merge()` — Unión tipo SQL (por columnas)

Combina DataFrames usando claves (columnas), similar a SQL.

Tipos: `inner`, `left`, `right`, `outer`.

```python
import pandas as pd

df_clientes = pd.DataFrame({
    'id': [1, 2, 3],
    'nombre': ['Ana', 'Luis', 'Pedro']
})

df_ventas = pd.DataFrame({
    'id': [1, 2, 4],
    'monto': [100, 200, 300]
})

inner_res = pd.merge(df_clientes, df_ventas, how='inner', on='id')
left_res = pd.merge(df_clientes, df_ventas, how='left', on='id')
outer_res = pd.merge(df_clientes, df_ventas, how='outer', on='id')

inner_res, left_res, outer_res
```

## 🔷 2. `join()` — Unión por índice

Simplifica uniones basadas en índices.

```python
df1 = pd.DataFrame({'A': [1, 2, 3]}, index=['a', 'b', 'c'])
df2 = pd.DataFrame({'B': [4, 5, 6]}, index=['a', 'b', 'd'])

inner_join = df1.join(df2, how='inner')
outer_join = df1.join(df2, how='outer')

inner_join, outer_join
```

## 🔷 3. `concat()` — Concatenación

Apila DataFrames por filas o columnas.

```python
df1 = pd.DataFrame({'A': [1, 2]})
df2 = pd.DataFrame({'A': [3, 4]})

concat_rows = pd.concat([df1, df2])
concat_cols = pd.concat([df1, df2], axis=1)

concat_rows, concat_cols
```
