```python
# Ejemplos de operadores relacionales

# Asignamos valores a las variables
a = 15
b = 3.5

# Comparaciones y resultados booleanos
print("¿a es igual a b?: ", (a == b))
print("¿a es distinto de b?: ",a != b)
print("¿a es mayor que b?: ",a > b)
print("¿a es menor que b?: ",a < b)
print("¿a es mayor o igual que b?: ",a >= b)
print("¿a es menor o igual que b?:",a <= b)
```

```python
# Ejemplos de uso de operadores relacionales y booleanos

# Asignamos valores a las variables
a = 5.3
b = 15
c = 9.78 

# Comparaciones y combinaciones de condiciones
print("¿a es igual a b y c es mayor que a?: ", a == b and c > a) # No hacen falta los parentesis
print("¿a es menor que b y c es menor que b?: ", (a < b and c < b))
print("¿a es menor que b o c es mayor que b?: ", (a < b or c > b))
print("¿a es distinto de c y c es mayor o igual que b?: ", (a != c and c >= b))
print("¿c está entre a y b?: ", (a < c and c < b))  # c debe estar entre a y b
```

Por ejemplo, para la primera comparacion del cuadro anterior, tenemos que **a = b** -> Falso, **c > a** -> Verdadero. Al consultar la tabla previa vemos que **False and True = False**, que es el resultado que arroja Python.
El mismo analisis se debe hacer con cualquiera de las otras comparaciones.

### **2b) Operadores de asignación**

Es muy frecuente que en el contrexto de un código sea necesario asignar un nuevo valor a una variable que tenía contenido.
```python
x = 4
x = x + 2
print(x)
```

```python
x = 4
x += 2  # en lugar de x = x + 2
print(x)

# Otros ejemplos

y = 7
y *= 2  # en lugar de  y = 2 * y
print(y)

z = 5
z **= 2 # en lugar de z = z**2
print(z)
```

### **Cómo utilizamos la funciones que creamos?**

Ejemplo:

```python
# Paso 1: Definimos la funcion (suma 3 a un numero dado) 

def add_three(input_var):
    output_var = input_var + 3
    return output_var

# Paso 2: utilizamos la funcion

new_number = add_three(4)    #<- El argumento (input_var) vale 4 en este caso

# Revisamos el resultado (7 en este caso)
print(new_number)
```

### **Alcance *local* de las variables definidas dentro de una función**

Modifiquemos levemente el ejemplo anterior:


```python
# Paso 1: Definimos la funcion (suma 3 a un numero dado) PERO SIN RETURN

def add_three_modif(input_var):
    output_var = input_var + 3
    # No hay return, por lo que no devuelve nada

# Paso 2: utilizamos la funcion
new_number_modif = add_three_modif(4)  #<- El argumento (input_var) vale 4 en este caso

# Revisamos el resultado (7 en este caso)       
print(new_number_modif)  # Esto imprimirá 'None'   
```

- La función no devuelve nada. Esto es porque la variable 'output_var' es local a la función y no se habilita su contenido con **return** para ser utilizado fuera de la misma. Por lo tanto, si se quiere tener disponible el contenido de una variable en el ambiente externo al de la función, debemos utilizar **return**.

### **Funciones *con un número arbitrario de argumentos* de entrada**

Si no sabes cuántos argumentos se pasarán a tu función, coloca un asterisco (*) antes del nombre del parámetro en la definición de la función. De esta manera, la función recibirá todos los argumentos como una **tupla**, y podrás acceder a cada elemento según sea necesario. 

Los argumentos arbitrarios suelen abreviarse como *args en la documentación de Python.

Ejemplo:

```python
def sumar_todos(*args):
    return sum(args)  # args es una tupla

print(sumar_todos(2, 4, 6))       # 12
print(sumar_todos(1, 2, 3, 4, 5)) # 15
print(sumar_todos())              # 0
```

- Que ocurre aquí: *args convierte todos los valores recibidos en una tupla. Esto permite iterar sobre ellos, sumarlos, contarlos, entre muchas otras operaciones, lo que lo hace muy flexible. En el ejemplo, la función muestra la suma de 3 números, de 5 números y también el resultado cuando no se pasa ningún argumento. Es muy común en Python encontrar funciones definidas de esta manera en librerías y paquetes.

### **Funciones con *argumentos clave-valor* de entrada (*keyword arguments*)**

También es posible pasar argumentos indicando el nombre del parámetro seguido de su valor (clave = valor).
Esto permite que el orden de los argumentos en la llamada a la función sea irrelevante. Muy útil para no recordar el orden específico de los parámetros dentro de una función.

Los *keyword arguments* suelen abreviarse como **kwargs** en la documentación de Python.

Ejemplo:

```python
def mostrar_info(ciudad, pais, continente):
    print("La ciudad más visitada es " + ciudad)

mostrar_info(continente="Europa", pais="España", ciudad="Barcelona")
```

### **Funciones con *un número arbitrario argumentos clave-valor* de entrada**

Si no sabes cuántos argumentos con nombre (keyword arguments) se pasarán a tu función, coloca dos asteriscos (**) antes del nombre del parámetro en la definición de la función.
De esta manera, la función recibirá todos esos argumentos en un diccionario, y podrás acceder a cada elemento utilizando su clave correspondiente.

Los argumentos de clave-valor arbitrarios suelen abreviarse como **kwargs en la documentación de Python.

```python
def mostrar_info(**kwargs):
    print("Tipo de kwargs:", type(kwargs))
    print("Contenido:", kwargs)

mostrar_info(nombre="Ana", edad=30, ciudad="Madrid")
mostrar_info(nombre="Carlos", edad=55)
```

Además de las funciones, en Python existen **métodos**, que son funciones asociadas a tipos de datos específicos (como diccionarios, listas, strings o DataFrames).

Listaremos a continuación algunos métodos aplicados a listas y diccionarios.

- Ejemplos: A continuación mostramos algunos usos de métodos con listas:


```python
mi_lista = [3, 1, 2]

mi_lista.append(4)        # [3, 1, 2, 4]
mi_lista.sort()           # [1, 2, 3, 4]
mi_lista.insert(0, 0)     # [0, 1, 2, 3, 4]
mi_lista.pop()            # [0, 1, 2, 3]
mi_lista.remove(2)        # [0, 1, 3]
print(mi_lista)
```

- Ejemplos: A continuación mostramos algunos usos de métodos con diccionarios:

```python
persona = {"nombre": "Ana", "edad": 25}

print(persona.get("nombre"))            # Ana
print(persona.get("altura", 1.70))      # 1.70 (valor por defecto)

persona["edad"] = 26
persona.update({"profesion": "Ingeniera"})

print(persona.keys())                   # dict_keys(['nombre', 'edad', 'profesion'])
print(persona.values())                 # dict_values(['Ana', 26, 'Ingeniera'])

persona.pop("edad")                     # Elimina la clave 'edad'
persona.setdefault("pais", "Argentina") # Agrega si no existe
print(persona)
```
