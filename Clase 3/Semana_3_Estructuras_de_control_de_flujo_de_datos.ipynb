## **1) Paquetes (librerías)**

Para el análisis de datos en Python, el uso de paquetes es fundamental, ya que extienden las capacidades del lenguaje base, que es de propósito general. Estos paquetes proporcionan herramientas especializadas para la manipulación, limpieza, análisis, modelado y visualización de datos, permitiendo un enfoque computacional y la realización de pruebas estadísticas con conjuntos de datos reales.


### **Algunos paquetes relevantes en Ciencia de Datos**

- *NumPy*: Es el paquete fundacional para la computación numérica en Python. Proporciona un objeto de arreglo multidimensional eficiente (ndarray) para operaciones aritméticas rápidas y vectorizadas.

- *pandas*: Ofrece estructuras de datos de alto nivel y herramientas diseñadas para trabajar con datos estructurados o tabulares de manera intuitiva y flexible. Es fundamental para la preparación, limpieza y análisis de datos.

- *Matplotlib*: Es la librería más popular para producir gráficos y otras visualizaciones de datos 2D en Python. Es configurable y útil para crear gráficos listos para publicación.

- *SciPy*: Colección de paquetes que abordan problemas fundamentales en la computación científica, ofreciendo herramientas para estadísticas, álgebra lineal, optimización, procesamiento de señales, entre otros.

- *scikit-learn*: Es el kit de herramientas de aprendizaje automático de propósito general para Python, incluyendo métodos supervisados y no supervisados, selección de modelos, evaluación y transformaciones de datos.

Esta lista no es exhaustiva, como es de esperar. A medida que vayamos utilizándolos iremos profundizando en algunos de ellos.

### **Uso de un gestor de instalación de paquetes**

En la clase anterior, vimos que para instalarlos es necesario el uso de un gestor. Aunque hay varias maneras de instalar paquetes en Pytnon, **pip** es la opcion mas simple. Se puede usar desde la terminal (como ya fue utilizado preciamente), o desde un Jupyter Notebook, por ejemplo, para instalar la libreria **numpy** desde una celda escribimos:

**!pip install numpy**

Si estuvieramos utilizando alguno de los ambientes **Anaconda** o **Miniconda**, el gestor es **conda**, y se utiliza así para la misma tarea:

**!conda install numpy**

*Nota: **Anaconda** es una distribucion de Python (y R) que incluye intérprete, gestor de paquetes (conda) y herramientas de desarrollo como Jupyter Lab, Navigator, Spyder (entorno de programación similar a VSC).
Trae preinstalados una gran variedad de paquetes de uso común, por lo que excepto algunas excepciones, no es necesario instalarlos. Aunque esta es una opción muy cómoda, insume muchos recursos*
