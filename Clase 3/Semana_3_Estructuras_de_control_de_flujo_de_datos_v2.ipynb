## 3.1) Particularidades del lenguaje

La elección de Python de utilizar la **sangría (indentación)** en lugar de llaves (${}$) es una de las características más distintivas del lenguaje y forma parte de su filosofía de diseño, que prioriza la **legibilidad y la simplicidad**.

A continuación, se detalla cómo funciona este sistema y las reglas que lo rigen:

### a. El inicio de un bloque: Los dos puntos (`:`)

En lenguajes como C++ o Java, un bloque de código comienza y termina con llaves. En Python, un bloque de código se introduce mediante el símbolo de **dos puntos (`:`)**. Este símbolo indica que la siguiente línea (y todas las consecutivas con el mismo nivel de sangría) constituye un nuevo cuerpo de instrucciones, ya sea para una función, un condicional (`if`), o un bucle (`for`, `while`).

### b. Estructura mediante espacios en blanco

Una vez que se inicia un bloque tras los dos puntos, el código debe estar desplazado hacia la derecha. La regla fundamental es que **todo el código dentro de un mismo bloque debe tener exactamente la misma cantidad de sangría** hasta que dicho bloque finalice. Si el nivel de sangría cambia hacia la izquierda, Python interpreta que el bloque actual ha terminado.

### c. Estándares y mejores prácticas

Aunque técnicamente se pueden utilizar tanto espacios como tabulaciones para indentar, se recomienda encarecidamente seguir estas pautas:

- **Cuatro espacios:** Utilizar **cuatro espacios** como el estándar de sangría por defecto.
- **Evitar tabulaciones:** Muchos programadores configuran sus editores de texto para que la tecla "Tab" inserte automáticamente cuatro espacios en su lugar, evitando mezclar ambos tipos de caracteres.
- **Automatización:** Entornos como **Jupyter Notebook e IPython** están diseñados para facilitar esto, insertando automáticamente el nivel de sangría correcto en la línea siguiente tras detectar un símbolo de dos puntos.

### d. Diferencias clave con otros lenguajes

- **Sin llaves:** Python elimina la necesidad visual de las llaves para agrupar sentencias, lo que a menudo hace que el código parezca más limpio y similar al "pseudocódigo".
- **Sin puntos y comas obligatorios:** A diferencia de otros lenguajes donde cada sentencia debe terminar en `;`, en Python las sentencias terminan con el salto de línea. Aunque se pueden usar puntos y comas para separar varias instrucciones en una misma línea, es una práctica que se desaconseja porque reduce la legibilidad del código.
- **Uso de `pass`:** Debido a que la sangría es obligatoria para definir bloques, Python incluye la palabra clave **`pass`**. Esta actúa como una instrucción nula ("no hace nada") y se utiliza en bloques donde la sintaxis exige código indentado pero no se requiere realizar ninguna acción específica en ese momento.

### e. Beneficios del sistema

Este sistema de "sangría significativa" obliga al programador a mantener una **estructura visual coherente**. Al ser el espacio en blanco una parte integral de la sintaxis y no solo una cuestión estética, se garantiza que la estructura lógica del programa coincida siempre con su apariencia visual, lo que facilita enormemente la revisión y el mantenimiento del código por parte de otros desarrolladores.

## 3.2) Estructuras Condicionales (`if`, `elif`, `else`)

Permiten ejecutar diferentes bloques de código según condiciones lógicas.
En este caso, los dos puntos **(:)** abren el bloque y las líneas siguientes deben estar alineadas hacia la derecha para que Python sepa qué instrucciones pertenecen a cada condición

````python
# Ejemplo 1
x = 10

if x > 0:
    print("x es positivo")
elif x == 0:
    print("x es cero")
else:
    print("x es negativo")
````

````python

# Ejemplo 2
z = 9
if z % 2 == 0:
    # Este es el bloque del 'if'
    print("z es divisible por 2")
elif z % 3 == 0:
    # Este es el bloque del 'elif'
    print("z es divisible por 3")
else:
    # Este es el bloque del 'else'
    print("z no es divisible por 2 ni por 3")
````

## 3.4) Bucle `for`

Se utiliza para iterar sobre secuencias. Aquí, todo lo que se repite en cada iteración del bucle debe estar indentado. Si una línea vuelve al margen izquierdo, Python interpreta que el bucle ha terminado.

El Bucle for en Python
El bucle for es una de las herramientas más potentes y utilizadas en Python para el control de flujo. A diferencia de otros lenguajes donde los bucles suelen basarse en contadores numéricos, el for de Python está diseñado específicamente para la iteración sobre colecciones de datos.

### a. Concepto y Propósito Principal
El propósito principal de un bucle for es ejecutar un bloque de código una vez por cada elemento presente en una secuencia o en un objeto ejecutable (iterador). Es la estructura ideal cuando:

- Sabes de antemano cuántas veces necesitas repetir una acción.

- Tienes una colección definida (lista, tupla, diccionario) que procesar.

### b. Sintaxis y Estructura
La sintaxis estándar en Python es la siguiente:

```
for variable in secuencia:
    # Bloque de código que se ejecuta con cada elemento
    print(variable)
```
La variable: Toma el valor del elemento actual de la secuencia en cada vuelta del bucle.

Los dos puntos `(:)`: Indican el inicio del cuerpo del bucle.

La sangría (indentación): Todo el código que debe repetirse debe estar desplazado a la derecha. Python utiliza este espacio en blanco para delimitar el bloque.

### c. Iteración sobre diferentes estructuras
El bucle for permite recorrer diversos tipos de datos de forma nativa:

- Listas y Tuplas: Recorre cada elemento en su orden original.

- Cadenas de texto (Strings): Itera sobre cada carácter individual.

- Diccionarios: Por defecto itera sobre las claves. Para obtener clave y valor simultáneamente, se usa `.items()`.

- NumPy Arrays: Permite procesar elementos de arreglos multidimensionales.

- Pandas DataFrames: Se puede iterar sobre filas mediante métodos como `.iterrows()`. Abordaremos este método más adelante cuando trabajemos con la libreria `pandas`.

### d. Herramientas complementarias útiles
Existen funciones integradas que potencian el uso del bucle:

`range()`: Genera una secuencia de números enteros (ej. range(10) para ir de 0 a 9).

`enumerate()`: Devuelve tanto el índice (posición) como el valor del elemento actual.

Desempaquetado (Unpacking): Si la colección contiene sub-secuencias, puedes asignar valores a múltiples variables directamente.

### d. Control del flujo dentro del bucle
Puedes alterar la ejecución mediante palabras clave:

`continue`: Salta el resto del código en la iteración actual y pasa inmediatamente al siguiente elemento.

`break`: Detiene y sale por completo del bucle de forma prematura.

### e. Rendimiento: Bucle for vs. Vectorización
Aunque el bucle for es fundamental, en contextos de Ciencia de Datos con grandes volúmenes de información, las operaciones vectorizadas (de librerías como NumPy o Pandas) son significativamente más rápidas.

#### Nota: La vectorización procesa bloques enteros de datos a la vez en memoria de bajo nivel, evitando la sobrecarga de procesar elemento por elemento que requiere un bucle for tradicional.

````python
# Ejemplo 1

for i in range(5):
    print(i)
````

````python
# Ejemplo 2

fam = [1.73, 1.68, 1.71, 1.89]
for height in fam:
    # Todo este bloque pertenece al bucle 'for'
    print(height)
# Esta línea ya está fuera del bucle porque no tiene sangría
print("Fin de la lista")
````

## **Ejemplos de Iteración sobre diferentes estructuras**

El bucle `for` de Python es extremadamente flexible y puede recorrer diversas estructuras de datos nativas y de librerías externas:

-   **Listas:** Es el uso más común. Por ejemplo, si tenemos una lista de alturas `fam = [1.73, 1.68, 1.71, 1.89]`, podemos imprimir cada valor con `for height in fam: print(height)`.

-   **Cadenas de texto (Strings):** Las fuentes indican que un string se trata como una secuencia de caracteres. Un ejemplo es `for c in "family": print(c.capitalize())`, que procesará cada letra individualmente.

-   **Diccionarios:** Para obtener tanto la clave como el valor, se debe utilizar el método `.items()`. Un ejemplo sería: `for key, value in world.items(): print(key + " -- " + str(value))` 

-   **NumPy Arrays:** En un array de una dimensión, funciona igual que una lista \[Source: 5: intermediate\_python\_chapter4.pdf\] . Sin embargo, en arrays de dos dimensiones (matrices), el bucle `for` recorre las **filas** por defecto \[Source: 6: intermediate\_python\_chapter4.pdf\] . Si se desea iterar sobre cada elemento escalar individual, se utiliza la función `np.nditer(my_array)` 

-   **Pandas DataFrames:** Si se itera directamente sobre un DataFrame, Python recorrerá los **nombres de las columnas** . Para iterar sobre las **filas**, la herramienta estándar es el método `.iterrows()`, que devuelve el índice (etiqueta) y los datos de la fila en cada iteración: `for lab, row in brics.iterrows(): print(lab + ": " + row["capital"])`.



## **Ejemplos de Herramientas complementarias útiles**

Estas funciones integradas permiten optimizar y dar mayor control a las iteraciones:

-   `range()` **:** Genera una secuencia de números enteros, muy útil cuando se conoce el número de repeticiones necesarias pero no se tiene una lista previa Ejemplo: `for i in range(5): print(i)` imprimirá los números del 0 al 4  . También puede generar pasos específicos, como `list(range(0, 20, 2))` para obtener solo números pares 

-   `enumerate()` **:** Resuelve el problema de necesitar el **índice** numérico además del valor del elemento . Ejemplo: `for index, height in enumerate(fam): print("index " + str(index) + ": " + str(height))`.

-   **Desempaquetado (Unpacking):** Permite asignar los elementos de una secuencia interna a variables individuales en la misma declaración del `for` Por ejemplo, si tenemos una lista de tuplas `seq = [(1, 2, 3), (4, 5, 6)]`, podemos iterar como `for a, b, c in seq: print(f'a={a}, b={b}, c={c}')`.



## 3.5) Bucle `while`

Se ejecuta mientras una condición sea verdadera.
El bucle `while` es una de las estructuras fundamentales para el control de flujo en Python, y su funcionamiento se basa exclusivamente en la evaluación continua de una **condición booleana**.

A continuación, te explico detalladamente cómo se comporta esta estructura y qué factores determinan su finalización:

### a. Mecanismo de Evaluación Continua

A diferencia de un condicional `if`, que evalúa una condición una sola vez, el bucle `while` funciona como un **"if repetido"**. El proceso sigue estos pasos lógicos:

**Evaluación inicial:** Antes de entrar al bucle, Python revisa la condición.

**Ejecución del bloque:** Si el resultado es `True`, se ejecutan todas las instrucciones indentadas dentro del bucle.

**Retorno al inicio:** Una vez terminada la última línea del bloque, el intérprete vuelve a la línea del `while` y evalúa la condición nuevamente.

**Terminación:** En el instante en que la condición se evalúa como `False`, el bloque de código se omite por completo y el programa continúa con la siguiente instrucción que no tenga sangría.

### b. Ejemplo práctico de terminación

Considera un caso donde queremos reducir un error hasta que sea insignificante:

```
error = 50.0
while error > 1:
    print(error)
    error = error / 4
```

**Primera vuelta:**`50.0 > 1` es `True`. El error pasa a ser `12.5`.

**Segunda vuelta:**`12.5 > 1` es `True`. El error pasa a ser `3.125`.

**Tercera vuelta:**`3.125 > 1` es `True`. El error pasa a ser `0.78125`.

**Cuarta evaluación:**`0.78125 > 1` es `False`. El bucle se detiene inmediatamente y no vuelve a imprimir nada.

### c. Otras formas de finalizar un bucle `while`

Aunque la condición booleana es el método estándar, las fuentes mencionan otra herramienta crítica para forzar la salida:

**La sentencia** `break`**:** Esta palabra clave permite **interrumpir y salir del bucle prematuramente**, incluso si la condición principal todavía es verdadera. Es extremadamente útil cuando el motivo para detenerse depende de un evento que ocurre dentro del bloque de código.

### d. El peligro de los bucles infinitos

Para que un bucle `while` finalice correctamente, el código dentro del bloque **debe realizar alguna acción que eventualmente cambie el resultado de la condición a** `False`.

Si la condición siempre es verdadera (por ejemplo, si olvidas actualizar la variable que se está comparando), el programa entrará en un **bucle infinito**.

En sistemas locales, esto suele requerir una interrupción manual mediante comandos como **Control + C** para forzar la detención del proceso.

### e. Resumen de diferencias clave: `while` vs. `for`

Es útil contrastar el `while` con el bucle `for` para entender su propósito:

El `for` se utiliza habitualmente cuando **ya se conoce el número de iteraciones** o se recorre una colección finita (como una lista).

El `while` se prefiere cuando **la duración depende de una condición externa o dinámica** que puede cambiar en cualquier momento durante la ejecución.

````python
# Ejemplo 1

contador = 0

while contador < 5:
    print(contador)
    contador += 1
````

## **Otros ejemplos de uso del lazo `while`**

Para complementar la explicación sobre el funcionamiento y la terminación de los bucles `while`, aquí tienes varios ejemplos prácticos extraídos de las fuentes que ilustran diferentes situaciones de uso:

### a. Ejemplo numérico: Reducción de un valor

Este es el ejemplo clásico de un cálculo que se repite hasta que una variable cae por debajo de un umbral específico.

```
error = 50.0
# El bucle se ejecutará mientras el error sea mayor que 1
while error > 1:
    error = error / 4
    print(error)
```
**Explicación:** El bucle comienza con un error de `50.0`. En cada vuelta, divide el valor por 4. Cuando el valor llega a `0.78125`, la condición `error > 1` se vuelve **falsa** y el bucle termina automáticamente.

### b. Ejemplo de simulación: Lanzar una moneda hasta que salga cara

Este ejemplo utiliza un bucle para repetir un experimento aleatorio cuya duración no conocemos de antemano.

```
import numpy.random as npr

intentos = 0
while True:  # Bucle infinito controlado manualmente
    moneda = npr.randint(2) # Genera 0 o 1
    intentos += 1
    if moneda == 1: # Supongamos que 1 es 'Cara'
        break # La sentencia break fuerza la salida inmediata
print(f"Salió cara en el intento {intentos}")
```
**Explicación:** Aquí se utiliza `while True` para crear un bucle que, en teoría, podría durar para siempre. Sin embargo, la instrucción `break` permite interrumpirlo en el momento exacto en que se cumple la condición deseada (obtener cara).

### c. Ejemplo de interacción: Sumar números hasta ingresar cero

Este tipo de bucle es común en programas que dependen de la entrada del usuario para saber cuándo detenerse.

```
suma = 0
numero = -1 # Valor inicial para entrar al bucle

while numero != 0:
    numero = int(input("Ingrese un número para sumar (0 para salir): "))
    suma += numero

print(f"La suma total es: {suma}")
```
**Explicación:** El bucle verifica si el `numero` es distinto de cero. Mientras el usuario siga ingresando valores diferentes a `0`, el programa continuará acumulando la suma. Al ingresar `0`, la condición `numero != 0` resulta **falsa** y el programa avanza a la siguiente instrucción fuera del bloque.

### d. Ejemplo de contador: Repetición controlada

Aunque para contar se suele usar el bucle `for`, también se puede hacer con un `while` gestionando manualmente el incremento.

```
contador = 0
# Se ejecuta mientras el contador sea menor a 5
while contador < 5:
    print(contador)
    contador += 1 # Es vital incrementar para evitar un bucle infinito
```
**Explicación:** Si olvidáramos la línea `contador += 1`, el valor siempre sería `0`, la condición `0 < 5` siempre sería verdadera y entraríamos en un **bucle infinito**, lo cual requeriría forzar la detención del sistema (usualmente con **Control + C**).

### Diferencias clave observadas en los ejemplos:

**Condición Dinámica:** En el ejemplo del error y la moneda, no sabemos cuántas iteraciones habrá; depende de los cálculos o del azar.

**Uso de** `break`**:** Permite salir del bucle en cualquier punto de su ejecución, sin esperar a que se evalúe la condición del encabezado.

**Importancia de la Indentación:** Como se observa en todos los bloques, el código que se repite debe estar desplazado a la derecha. Si una línea vuelve al margen izquierdo, Python entiende que está fuera del bucle.


## **Ejemplos de Control del flujo dentro del bucle**

Estas palabras clave alteran el comportamiento normal de la iteración en respuesta a condiciones específicas:

-   **continue** **:** Se utiliza para **saltar el resto del código** en la iteración actual y pasar inmediatamente al siguiente elemento. Un ejemplo práctico es cuando se quiere sumar números en una lista pero ignorar los valores nulos (`None`).

-   **break :** Se utiliza para **salir completamente** del bucle de forma prematura Por ejemplo, si se busca un valor específico y se quiere dejar de iterar en cuanto se encuentra.

-   **pass** **:** Aunque no altera el flujo (no hace nada), es necesario como **marcador de posición** en bloques donde la sintaxis exige código indentado pero no se requiere realizar ninguna acción todavía.

````python
# Ejemplo continue
for i in range(5):
    if i == 2:
        continue
    print(i)
````

````python
# Ejemplo break
for i in range(10):
    if i == 5:
        break
    print(i)
````

Si por algún motivo necesitas declarar una estructura pero aún no quieres escribir código en ella, debes usar `pass` para que Python no dé un error de sintaxis por falta de un bloque indentado \[Source: 9: Wes McKinney - Python for Data Analysis\_ Data Wrangling with pandas, NumPy, and Jupyter-O'Reilly Media (2022).pdf\].

````python
# Ejemplo pass

if x < 0:
    print("¡Negativo!")
elif x == 0:
    # Python exige algo aquí tras los dos puntos
    pass 
else:
    print("¡Positivo!")
````

## **Ejemplo de bloques anidados (Bucle con `if` dentro)**

Cuando pones una estructura dentro de otra (como un `if` dentro de un `for`), cada nivel requiere un nivel adicional de sangría (normalmente otros 4 espacios más).

````python
# Ejemplo 1 bloques anidados

for i in range(3):
    for j in range(2):
        print(i, j)
````

````python
# Ejemplo 2 bloques anidados

import random # Importamos libreria random para generar números aleatorios

num_sims = 100 # Numero de simulaciones a realizar
threshold = 6 # Umbral para contar el número de caras (H) en cada simulación
for sim in range(num_sims):
    # Primer nivel de indentación (dentro del for)
    num_heads = random.choices(['H', 'T'], k=20).count('H')
    if num_heads <= threshold:
        # Segundo nivel de indentación (dentro del if, que está dentro del for)
        print(f'Simulación {sim}: {num_heads} caras')
````
