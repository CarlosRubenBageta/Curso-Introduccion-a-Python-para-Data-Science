```python
# NumPy is imported, seed is set
import numpy as np
np.random.seed(123)

# Starting step
step = 50

# Roll the dice
dice=np.random.randint(1,7)

# Finish the control construct
if dice <= 2 :
    step = step - 1
elif dice > 2 and dice < 6 :
    step += 1
else :
    step = step + np.random.randint(1,7)

# Print out dice and step
print(dice, step)
```

**The next step**

```python
# NumPy is imported, seed is set
import numpy as np

# Initialize random_walk
random_walk=[0]
np.random.seed(123)

# Complete the ___
for x in range(100) :
    # Set step: last element in random_walk
    step=random_walk[-1]

    # Roll the dice
    dice = np.random.randint(1,7)

    # Determine next step. 
    if dice <= 2:
        step = max(0,step - 1) #Arregla si random_walk es negativo
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)

    # append next_step to random_walk
    random_walk.append(step)

# Print random_walk
print(random_walk) 
```

```python
# Este codigo plotea el random_walk en función de la cantidad de veces que tiro el dado

import matplotlib.pyplot as plt

# Make a plot

numeros = [i for i in range(0, len(random_walk))] # Esta es una manera compacta de generar una lista incremental
plt.plot(numeros,random_walk)

# Display the plot
plt.show()
```

```python
# NumPy is imported, seed is set
import numpy as np

np.random.seed(123)

# Initialize all_walks (don't change this line)
all_walks = []

# Simulate random walk 10000 times
for i in range(500) :

    # Code from before
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)

        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        random_walk.append(step)

    # Append random_walk to all_walks
    all_walks.append(random_walk[-1])

# Print all_walks
print(all_walks)

# Este codigo plotea el random_walk en función de la cantidad de veces que tiro el dado

import matplotlib.pyplot as plt

# Make a plot

plt.hist(all_walks,bins=20)

# Display the plot
plt.show()
```

**Implement clumsiness -  Aquí implementamos la probabilidad de caerse**


```python
# numpy and matplotlib imported, seed set

# NumPy is imported, seed is set
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(123)

# Simulate random walk 250 times
all_walks = []
for i in range(250) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)

        # Implement clumsiness - Aqui implementamos la probabilidad de caerse
        if np.random.rand() <= 0.001 :
            step = 0

        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))
plt.plot(np_aw_t)
plt.show()
```

**Plot the distribution**

```python
# numpy and matplotlib imported, seed set

# NumPy is imported, seed is set
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(123)

# Simulate random walk 500 times
all_walks = []
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))

# Select last row from np_aw_t: ends
ends = np_aw_t[-1,:]

# Plot histogram of ends, display plot
plt.hist(ends,bins=20)
plt.show()
```
