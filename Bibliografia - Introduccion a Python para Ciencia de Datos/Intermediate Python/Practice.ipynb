```python
import pandas as pd
product=['A','A','A','B','B','B']
month=['Jan','Feb','Mar','Jan','Feb','Mar']
sold=[85,200,180,90,80,130]
sales=pd.DataFrame(product,month,sold)
print(sales)
```

```python
Following is a preview of the DataFrame sales:

product month sold
      A   Jan   85
      A   Feb  200
      A   Mar  180
      B   Jan   90
      B   Feb   80
      B   Mar  130 
```

```python
import numpy as np
pick = np.logical_or(sales['product'] == 'B',?)
sales[pick]
---
""""Required Output:
product month  sold
1       A   Feb   200
2       A   Mar   180
3       B   Jan    90
4       B   Feb    80
5       B   Mar   130
""""
```

```python
import numpy as np
pick = np.logical_or(sales['product'] == 'B',sales.loc['product'])
sales[pick]
---
```
