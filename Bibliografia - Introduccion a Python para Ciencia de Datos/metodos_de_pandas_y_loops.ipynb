```python
import pandas as pd

df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carlos'],
    'Edad': [23, 34, 45]
})


# Uso metodo iterrows()   - Itera sobre filas

for index, row in df.iterrows():
    print(f"Índice: {index}, Nombre: {row['Nombre']}, Edad: {row['Edad']}")
```

```python
# Uso metodo itertuples()  - Itera sobre filas

for row in df.itertuples():
    print(f"Índice: {row.Index}, Nombre: {row.Nombre}, Edad: {row.Edad}")
```

```python
# Uso metodo items()  - Itera sobre columnas

for nombre_columna, columna in df.items():
    print(f'Nombre columna:', {nombre_columna})
    print(f'Serie columna:', columna)
```
