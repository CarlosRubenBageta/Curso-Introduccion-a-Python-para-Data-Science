### **1 c) Manejo de entornos virtuales (Python)**

Un entorno virtual es una instancia aislada del intérprete de Python, junto con su propio conjunto de librerías y dependencias. Se crea y gestiona de manera local a nivel de proyecto. **No afecta al sistema operativo ni a otras instalaciones de Python**. Es necesario utilizar entornos virtuales?

- **Aislamiento de dependencias:**
Permite usar versiones específicas de librerías para cada proyecto, sin interferencia con otros.

Ejemplo: Un proyecto usa pandas==1.5 y otro pandas==2.2. 

- **Reproducibilidad:**
Al usar un entorno virtual junto con un archivo *requirements.txt* o *environment.yml*, puede reconstruir el mismo entorno en otra máquina o para otro colaborador con exactitud.

- **Evita conflictos con las distintas versiones de Python instaladas en el sistema:**
Especialmente en Windows, donde múltiples versiones de Python pueden convivir, los entornos virtuales evitan modificaciones peligrosas en la instalación global.

- **Mejor práctica para ciencia de datos, machine learning y desarrollo web:**
Librerías como scikit-learn, tensorflow, matplotlib, etc., tienen dependencias pesadas. Gestionarlas sin un entorno virtual puede generar errores difíciles de depurar.

**Scripts en Python... que son?**

Un **script** es un archivo de texto que contiene un conjunto de instrucciones que pueden ser ejecutadas por un **intérprete**. A diferencia de un programa compilado (como los escritos en C o Java), un script es interpretado línea por línea. En el caso de Python, estos scripts tienen extensión **.py**.

Desde el punto de vista práctico: un script de Python es simplemente un archivo que contiene código en lenguaje Python, y que se puede ejecutar para realizar tareas específicas. 

Podemos utilizar la terminal (consola) para ejecutar multiples lineas de comando **una por una**, pero esto probablemente no seria practico desde lo operativo si necesito ejecutar multiples comandos (lo que es comun). Por lo tanto, este tipo de tareas se ejecutan en scripts. Al ejecutar un script en Python (o cualquier lenguaje de programacion), se llevan a cabo todas las tareas una por una, de manera **automatica**. No se ejecutan las tareas cada una por separado, como ocurre cuando usamos la terminal.

**¿Qué puede contener un script de Python?**

Iremos ampliando muchas de estas tareas posteriormente, pero para ir fijando ideas, un script puede incluir:

- Declaraciones de variables

- Operaciones matemáticas

- Definición de funciones

- Llamadas a bibliotecas externas (por ejemplo pandas, matplotlib, numpy)

- Estructuras de control (condicionales, bucles)

- Lectura o escritura de archivos

- Conexión con bases de datos

- Ejecución de modelos de machine learning

- Automatización de tareas, entre otros.

**¿Como comienzo a trabajar con scripts?**

- EL primer paso es crear un archivo de texto (puede ser con cualquier editor, por ejemplo, con el bloc de notas de Windows).

- El segundo paso es escribir el contenido del script (lo que constituye el cuerpo del programa)

- Luego de finalizar, grabamos el script como : **nombre _archivo.py**, donde es imprescindible que la esxtension sea **.py**

**¿Como ejecutamos scripts?**

- Desde la terminal, podemos escribir: **python.exe nombre_archivo.py**. El script se deberia ejecutar con normalidad. Tambien deberia funcionar si escribimos **py nombre_archivo.py**.

**¿Que son variables en python?**

- En Python, una **variable** es un nombre simbólico que se usa para almacenar un valor y poder usarlo más adelante en un programa. Es como una caja con una etiqueta: le ponés un nombre a la caja (la variable) y guardás algo adentro (el valor).
